```python
n = input("Enter the number:")   # take input as string, not int
i = len(n) - 1                   # start from the last index
rev = ""                         # empty string to store reversed number

while i >= 0:
    rev = rev + n[i]             # add each digit from end
    i -= 1                       # move to the previous character

print("Reversed number:", rev)
```

```python
#4
n = int(input("Enter the number:"))

temp = n
rev = 0
while n>0:
    r = n%10
    n = n//10
    rev = rev*10 + r

if temp == rev:
    print("Palindrome")
else:
    print("Not")
```
